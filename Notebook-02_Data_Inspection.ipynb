# Data Inspection

### What is Data Inspection?

Data inspection is the process of examining a dataset to understand its structure, contents, data types, distributions, and potential problems before performing data analysis or machine learning.

Real-world example:

A telecom company receives customer data containing:

* Customer ID
* Age
* Contract type
* Monthly charges
* Internet service
* Churn status

Before building a churn prediction model, we need to inspect the data and answer questions such as:

* How many customers are there?
* Which columns are numerical?
* Which are categorical?
* Are there missing values?
* What values occur in each categorical column?

```python
import pandas as pd

df = pd.read_csv("customer_churn.csv")
df.head()
```

### 1. Number of Rows

The number of rows represents the number of records or observations in the dataset.

In this dataset, each row represents one customer record.

Real-world example:

If a telecom company has information about 5,020 customers, there will be 5,020 rows.

#### AI/ML usage

The number of observations affects:

* Model training
* Model accuracy
* Statistical reliability
* Train/test splitting

More useful observations generally give an ML model more information to learn from.

```python
print("Number of rows:", len(df))
```

### 2. Number of Columns

The number of columns represents the number of variables or attributes in the dataset.

Real-world example:

For a customer, we have information such as:

* Age
* Gender
* Monthly charges
* Contract type
* Internet service
* Churn

Each of these is represented by a column.

#### AI/ML usage

Columns usually become features used by an ML model.

```python
print("Number of columns:", len(df.columns))
```

```python
print(df.columns)
```

### 3. DataTypes

A data type tells us what kind of data is stored in a column.

Common types include:

| Data Type  | Meaning                       | Example      |
| ---------- | ----------------------------- | ------------ |
| `int64`    | Integer                       | 5, 10, 20    |
| `float64`  | Decimal number                | 25.5, 100.75 |
| `object`   | Usually text/categorical data | Male, Female |

#### AI/ML usage

Data types determine what preprocessing is required.

```python
df.dtypes
```

### 4. Index

An index is used to identify the position of each row in a Pandas DataFrame.

By default, Pandas creates a numerical index starting from 0.

Real-world example:

The index can be thought of as the row number:

Index     CustomerID

0         CUST00001

1         CUST00002

2         CUST00003

#### AI/ML usage

The index normally isn't used as an ML feature.

For example, this:

0, 1, 2, 3, 4...doesn't contain meaningful information about customer churn.

```python
print(df.index)
```

Why use it?

It helps us quickly understand:

Real-world usage:

If you have thousands of customer records, you don't want to print everything.

Instead, head() lets you quickly inspect the beginning of the dataset.

#### AI/ML usage

Useful for:

* Checking whether data loaded correctly
* Checking column names
* Checking sample values
* Detecting obvious formatting problems

### 13. value_counts()

value_counts() is a Pandas function used to count how many times each unique value appears in a column.

It is mainly useful for understanding the distribution of categorical data.

Example:

Suppose a telecom company has:

| Contract Type | Customers |
| ------------- | --------: |
| Monthly       |     2,500 |
| Yearly        |     1,500 |
| Two Year      |     1,020 |


This tells the company how many customers have each type of contract.

#### AI/ML usage

value_counts() is useful during Exploratory Data Analysis (EDA) to:

* Understand categorical distributions
* Identify class imbalance
* Check target variable distribution
* Find rare categories
* Detect unexpected values

```python
# For Gender:

df["Gender"].value_counts()
```

```python
# For Churn:

df["Churn"].value_counts()
```

### 17. Identifier Columns

An identifier column uniquely identifies each record or entity.

Examples:

* CustomerID
* EmployeeID
* OrderID
* TransactionID

An identifier is generally not a meaningful predictive feature.

Example:

A bank might have:

CustomerID = CUST001

CustomerID = CUST002

CustomerID = CUST003

The ID identifies the customer but doesn't describe their behavior.

#### AI/ML usage

Identifiers are generally excluded from ML models because they don't provide meaningful information.

Including them can:

* Add noise
* Cause overfitting
* Create misleading patterns
* Reduce model generalization

The target variable is the variable that we want our ML model to predict.

It is also called:

* Dependent variable
* Output variable
* Response variable
* Label

Example:

If a company wants to predict whether a customer will leave:

Churn is the target.

If Churn = Yes -> the customer left.

If Churn = No -> the customer stayed.

#### AI/ML usage
